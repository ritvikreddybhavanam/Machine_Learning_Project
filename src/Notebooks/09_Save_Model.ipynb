```
import pandas as pd
import joblib
import os

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np
```

```
train_path = "../../Dataset/processed/train_data.csv"
test_path = "../../Dataset/processed/test_data.csv"

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

print("Training data loaded successfully")
print("Testing data loaded successfully")

print("Training shape:", train_data.shape)
print("Testing shape:", test_data.shape)
```

```
X_train = train_data.drop(columns=["Revenue(INR)"])
y_train = train_data["Revenue(INR)"]

X_test = test_data.drop(columns=["Revenue(INR)"])
y_test = test_data["Revenue(INR)"]

print("Number of input features:", X_train.shape[1])
print("Target column:", "Revenue(INR)")
```

```
model_path = "../models/elastic_net_model.pkl"

if os.path.exists(model_path):
    print("Model file found")
    print("Model path: ", model_path)
else:
    print("Model file not found")
```

```
model = joblib.load(model_path)

print("Elastic Net model loaded successfully")
print(model)
```

```
y_pred = model.predict(X_test)

print("Predictions successful")
print("Number of predictions: ", len(y_pred))
```

```
prediction_data = pd.DataFrame({
    "Actual Revenue": y_test.values,
    "Predicted Revenue": y_pred
})

prediction_data.head(10)
```

```
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("MAE :", round(mae, 2))
print("MSE :", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 4))
```

```
model_info = {
    "model_name": "Elastic Net Regression",
    "target": "Revenue(INR)",
    "number_of_features": X_train.shape[1],
    "training_samples": X_train.shape[0],
    "testing_samples": X_test.shape[0],
    "alpha": model.named_steps["elastic_net"].alpha,
    "l1_ratio": model.named_steps["elastic_net"].l1_ratio,
    "mae": mae,
    "mse": mse,
    "rmse": rmse,
    "r2": r2
}

for key, value in model_info.items():
    print(key, ":", value)
```

```
model_info_df = pd.DataFrame(
    [model_info]
)

model_info_df.to_csv(
    "../../Dataset/processed/model_information.csv",
    index=False
)

print("Model information saved successfully")
```

```
final_model_path = "../models/final_elastic_net_model.pkl"

joblib.dump(
    model,
    final_model_path
)

print("Final model saved successfully")
print("Path:", final_model_path)
```

```
final_model = joblib.load(final_model_path)

print("Final model loaded successfully")
```

```
final_predictions = final_model.predict(X_test)

print("Final model prediction successful")
print("First 5 predictions:")
print(final_predictions[:5])
```

```
prediction_check = pd.DataFrame({
    "Original Prediction": y_pred[:5],
    "Final Model Prediction": final_predictions[:5]
})

prediction_check
```

```
print("==========================================")
print("       FINAL MODEL INFORMATION")
print("==========================================")

print("Model:", model_info["model_name"])
print("Target:", model_info["target"])
print("Features:", model_info["number_of_features"])
print("Training samples:", model_info["training_samples"])
print("Testing samples:", model_info["testing_samples"])

print("\nModel Parameters")
print("Alpha:", model_info["alpha"])
print("L1 Ratio:", model_info["l1_ratio"])

print("\nEvaluation")
print("MAE:", round(model_info["mae"], 2))
print("MSE:", round(model_info["mse"], 2))
print("RMSE:", round(model_info["rmse"], 2))
print("R²:", round(model_info["r2"], 4))

print("\nFinal model:")
print(final_model_path)
```
