```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = "../../Dataset/processed/cleaned_movie_data.csv"
```

```python
df = pd.read_csv(file_path)
print("Dataset loaded successfully")
print("Shape of the dataset: ", df.shape)
```

```python
df.head()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["Revenue(INR)"], bins=30, kde=True)

plt.title("Distribution of Revenue(INR)")
plt.xlabel("Revenue(INR)")
plt.ylabel("Number of Movies")

plt.savefig(
    "../output/plots/revenue_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["Budget(INR)"], bins=30, kde=True)

plt.title("Distribution of Budget(INR)")
plt.xlabel("Budget(INR)")
plt.ylabel("Number of Movies")

plt.savefig(
    "../output/plots/budget_vs_revenue.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["Number_of_Screens"], bins=30, kde=True)

plt.title("Distribution of Number of Screens")
plt.xlabel("Number of Screens")
plt.ylabel("Number of Movies")

plt.savefig(
    "../output/plots/2.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    y="Genre",
    order=df["Genre"].value_counts().index
)

plt.savefig(
    "../output/plots/3.png",
    dpi=300,
    bbox_inches="tight"
)

plt.title("Number of Movies by Genre")
plt.xlabel("Number of Movies")
plt.ylabel("Genre")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Release_Period",
    order=df["Release_Period"].value_counts().index
)

plt.savefig(
    "../output/plots/4.png",
    dpi=300,
    bbox_inches="tight"
)

plt.title("Movies by Release Period")
plt.xlabel("Release Period")
plt.ylabel("Number of Movies")

plt.xticks(rotation=45)

plt.show()
```

```python
numerical_columns = [
    "Number_of_Screens",
    "Revenue(INR)",
    "Budget(INR)"
]

correlation = df[numerical_columns].corr()

print(correlation)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.savefig(
    "../output/plots/10.png",
    dpi=300,
    bbox_inches="tight"
)

plt.title("Correlation Heatmap")

plt.show()
```
