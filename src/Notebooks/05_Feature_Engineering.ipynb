```python
import pandas as pd
import numpy as np
```

```python
file_path = "../../Dataset/processed/cleaned_movie_data.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully")
print("Shape:", df.shape)
```

```python
df = df.drop(columns=["Movie_Name"])

print("Movie_Name removed")
print("Shape: ", df.shape)
```

```python
X = df.drop(columns=["Revenue(INR)"])
y = df["Revenue(INR)"]

print("Features: ")
print(X.columns.tolist())

print("Target: ")
print("Revenue(INR)")
```

```python
categorical_columns = [
    "Release_Period",
    "Whether_Remake",
    "Whether_Franchise",
    "Genre",
    "New_Actor",
    "New_Director",
    "New_Music_Director",
    "Lead_Star",
    "Director",
    "Music_Director"
]

numerical_columns = [
    "Number_of_Screens",
    "Budget(INR)"
]

print("Categorical features:")
print(categorical_columns)

print("\nNumerical features:")
print(numerical_columns)
```

```python
for column in categorical_columns:
    print("\n" + column)
    print(df[column].unique())
```

```python
X_encoded = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True
)

print("Shape before encoding: ", X.shape)
print("Shape after encoding: ", X_encoded.shape)
```

```python
X_encoded = X_encoded.astype(int)

print(X_encoded.dtypes)
```

```python
print("Final feature shape:", X_encoded.shape)

X_encoded.head()
```

```python
print("Target shape:", y.shape)

print("\nTarget statistics:")
print(y.describe())
```

```python
print("Missing values in features:")
print(X_encoded.isnull().sum().sum())

print("\nMissing values in target:")
print(y.isnull().sum())
```

```python
final_dataset = X_encoded.copy()

final_dataset["Revenue(INR)"] = y.values

output_path = "../../Dataset/processed/final_dataset.csv"

final_dataset.to_csv(output_path, index=False)

print("Final dataset saved successfully")
print(output_path)
```
