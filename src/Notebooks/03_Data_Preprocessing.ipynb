```python
import pandas as pd
import numpy as np
```

```python
file_path = "../../Dataset/raw/Data_for_repository.csv"
```

```python
df = pd.read_csv(file_path)

print("Loaded dataset Successfully")
print("Shape of Dataset: ", df.shape)
```

```python
print("Missing values before preprocessing: ")
print(df.isnull().sum())
```

```python
print("Duplicated Records: ", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Shape of removing duplicates: ", df.shape)
```

```python
print(df.dtypes)
```

```python
categorical_columns = [
    "Release_Period",
    "Whether_Remake",
    "Whether_Franchise",
    "Genre",
    "New_Actor",
    "New_Director",
    "New_Music_Director"
]

for column in categorical_columns:
    print("\n", column)
    print(df[column].value_counts())
```

```python
numerical_columns = [
    "Number_of_Screens",
    "Revenue(INR)",
    "Budget(INR)"
]

for column in numerical_columns:
    df[column] = pd.to_numeric(df[column], errors='coerce')

print(df[numerical_columns].dtypes)
```

```python
for column in numerical_columns:
    print("\n", column)
    print("Negative Values: ", (df[column] < 0).sum())
```

```python
for column in numerical_columns:
    df.loc[df[column] < 0, column] = np.nan
```

```python
for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print("Missing values after preprocessing:")

print(df.isnull().sum())
```

```python
for column in numerical_columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outliers = ((df[column] < lower_limit) | (df[column] > upper_limit)).sum()

    print(column, ":", outliers, "outliers")
```

```python
output_path = "../../Dataset/processed/cleaned_movie_data.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully")
print(output_path)
```

```python
print("Final Shape: ", df.shape)

print("Missing Values: ")
print(df.isnull().sum())
print("\nDuplicate Records: ", df.duplicated().sum())
```
